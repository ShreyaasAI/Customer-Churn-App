```python
x=data.iloc[:,:-1]
y=data["Exited"]
y.value_counts()
```

```python
sns.boxplot(data=data["Age"])
```

```python
fig, ax = plt.subplots(figsize = (15, 15))
sns.boxplot(data = x, ax=ax)
plt.savefig('boxPlot.jpg')
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y, test_size=0.25,)
```

```python
x_train_resampled=x_train
y_train_resampled=y_train
# from imblearn.over_sampling import SMOTE
# smote = SMOTE(random_state=42)
# x_train_resampled,y_train_resampled=smote.fit_resample(x_train,y_train)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train_resampled)
x_scaled_test  =scaler.transform(x_test)
```

```python
x_train_scaled
```

```python
from sklearn.linear_model import LogisticRegression
LoRe = LogisticRegression()
LoRe.fit(x_train_scaled,y_train_resampled)
```

```python
print(LoRe.coef_)
print(LoRe.intercept_)
```

```python

y_pred=LoRe.predict(x_scaled_test)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
print(accuracy_score(y_test,y_pred))
```

```python
print(confusion_matrix(y_test,y_pred))
```

```python
temp = ["Retained","Churn"]
print(classification_report(y_test,y_pred,target_names=temp))
```

```python
from sklearn.ensemble import GradientBoostingClassifier
gbr = GradientBoostingClassifier(n_estimators = 200 , learning_rate=0.5, random_state = 41, max_features=9)
gbr.fit(x_train_scaled,y_train_resampled)
predicted_y = gbr.predict(x_scaled_test)
```

```python
accuracy_score(y_test,predicted_y)
```

```python
temp = ["Retained","Churn"]
print(classification_report(y_test,predicted_y,target_names=temp))
```
